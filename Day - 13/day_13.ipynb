```python
#  List comprehension 
# compact way of creating a lsit from a sequence. It is a short way to create a new list. List comprehension is considerably faster than processing a list using the for loop.
# syntax
# [expression for i in iterable if condition]
```

```python
# Change a string to a list of characters. 
# One way
language = 'Python'
lst = list(language) # changing the string to a list
print(type(lst))
print(lst)
```

```python
# Second way: list comprehension 
lst = [i for i in language]
print(type(lst))
print(lst)
```

```python
# Example 2 
# For instance if you want to generate a list of numbers. 

# Generating numbers 
numbers = [i for i in range(11)] # to generate numbers from 0 to 10 
print(numbers)

# it is possible to do mathematical operations during iteration
squares = [i * i for i in range(11)]
print(squares)

# Also possible to make list of tuples
numbers = [(i, i*i) for i in range(11)]
print(numbers)
```

```python
# Example 2 
# List comprehension can be combined with if expression 
# Generating even numbers 
even_numbers = [i for i in range(21) if i % 2 == 0] # to generate even numbers list in range 0 to 21
print(even_numbers)

# Generating odd numbers 
odd_numbers = [i for i in range(21) if i % 2 != 0] # generate odd numbers list in range 0 to 21
print(odd_numbers)

# Filter numbers, lets filter only positive even numbers 
# Filter numbers: let's filter out positive even numbers from the list below
numbers = [-8, -7, -3, -1, 0, 1, 3, 4, 5, 7, 6, 8, 10]
positive_even_numbers = [i for i in numbers if i % 2 == 0 and i > 0]
print(positive_even_numbers)     

# flattening a two dimensional array 
list_of_lists = [[1,2,3], [4,5,6], [7,8,9]]
flattening_list = [number for row in list_of_lists for number in row]
print(flattening_list)
```
