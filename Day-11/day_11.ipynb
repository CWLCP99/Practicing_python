```python
# Returning a list for example 
def find_even_numbers(n): 
    evens = []
    for i in range(n + 1): 
        if i % 2 == 0: 
            evens.append(i)
    return evens
print(find_even_numbers(10))

print(range(10))
```

```python
# If we don't know the number of arguments we pass to our function, we can create a function which can take arbitrary number of arguments by adding * before the parameter name. 
'''   
# syntax
# Declaring a function
def function_name(*args):
    codes
    codes
# Calling function
function_name(param1, param2, param3,..)
'''

def sum_all_nums(*nums):
    total = 0
    for num in nums:
        total += num     # same as total = total + num 
    return total
print(sum_all_nums(2, 3, 5)) # 10
```

```python
# Dictionary unpacking 
# We can call a function which has named arguments using a dictionary with matching key names. 
def greet(name, location): 
    # Print a greeting message using the provided arguments 
    print("Hi there", name, "how is the weather in", location)

# Call the function using keyword arguments 
greet(name= 'Alice', location = "New York")

# Create a dictionary with keys matching the functions parameter names
my_dict = {
    "name": "Alice", 
    "location": "New York"
    }

# Call the function using dictionary unpacking
greet(**my_dict)
```

```python
# Arbitrary number of arguments and sums all the arguments, check if the list items are number types. If not do a give a reasonable feedback
def add_all_nums(*nums): 
    total = 0 
    for i in nums:
        if isinstance(i, (int, float)) and not isinstance(i,bool): # True and False are technically  ints in Python, so True can count as 1.
            total += i
        else: 
            print(f"invalid number: {i!r} (type {type(i).__name__}, skipping")
    return total 
print(add_all_nums(1,2,3,5,8,23,3))
print(add_all_nums(1,"a"))
```

```python
# Level 2 
# 1. Declare a function named evens_and_odds. It takes positive integers as parameters and it counts number of evens and odds in the number. 
def evens_and_odds(n: int):
    if not isinstance(n, int) or  isinstance(n, bool) or n <= 0: 
        raise ValueError("n must be a positive integer")
    counter_even = 0 
    counter_odd = 0 
    for numbers in range(n+1): # range goes from 0 to n-1 so will need to add last value aswell. 
        if numbers % 2 == 0:
            counter_even += 1 
        else:
            counter_odd += 1 
    return counter_even, counter_odd
print(evens_and_odds(50))
    
```

```python
# Level 3 
# Write a function called is_prime, which checks if a number is prime. 
# prime number = whole number greater than 1 that can be divided by 1 or itself. 1 is not a prime number. 

# print( 7 % 1) --> 0 
# print(7 % 7) --> 0 

def is_prime(n): 
    if n < 2: 
        return ValueError(f"{n} is below 2. Not prime number")
    for candidate in range(2, n): 
        if n % candidate == 0: 
            return False
        
    return candidate
        
    
        
is_prime(3)

    


```
