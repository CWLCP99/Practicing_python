```python
# Statistics Module provide functions for mathematical statistics of numeric data. 
from statistics import * # importing all the statistics modules
ages = [20, 20, 4, 24, 25, 22, 26, 20, 23, 22, 26]
print(mean(ages))       # ~22.9
print(median(ages))     # 23
print(mode(ages))       # 20
print(stdev(ages))      # ~2.3
```

```python
# Math Module contains many mathematical operations and constraints. 
import math
print(math.pi)           # 3.141592653589793, pi constant
print(math.sqrt(2))      # 1.4142135623730951, square root
print(math.pow(2, 3))    # 8.0, exponential function
print(math.floor(9.81))  # 9, rounding to the lowest
print(math.ceil(9.81))   # 10, rounding to the highest
print(math.log10(100))   # 2, logarithm with 10 as base
```

```python
# Exercise: Level 1
# Write a function which generates a six digit/character random_user_id
import random 
import string 

# random.choice() picks one random item from a sequence (a list, a string, a tuple). 

# str.join() takes an iterable of strings (like a list) and glues them into one string, using the string you call it on as the separator. 
# separator.join(list)

# We'll need to call random.choice 6 six times and collect the results, one way to repeat something 6 times and collect the results is a loop that appends to a list, or a list comprehension. 

def random_user_id(): 
    # combine digits and letters 
    digits_letters_var = string.ascii_letters + string.digits
    
    # create a list 
    lst = []

    # for loop as we know that we need to run it 6 times only
    for i in range(6): 
        x = random.choice(digits_letters_var)
        lst.append(x)
    
    return ''.join(lst) 

random_user_id()
        
#print(random_user_id())
    
# digits_letters_var = string.ascii_letters + string.digits

# random.choice(digits_letters_var)



    


```
