```python
# To check the lenght of a string use the len() method. 
letter = 'P'
print(letter)
print(len(letter))
```

```python
# Multiline string is created using triple single (""") or triple double quotes ("""). 
multiline_string = '''I am a teach and I enjoy teaching. I didn't find anything as rewarding as empowering people.'''
print(multiline_string)

# Another way to do it
multiline_string = """I am a teach and I enjoy teaching. I didn't find anything as rewarding as empowering people."""
print(multiline_string)
```

```python
# String concatenation 
# Merging or connecting string is called concatenation 
first_name = 'Peter'
last_name = 'Garcia'

space = ' '

full_name = first_name + space + last_name
print(full_name)

# checking lenght 
print(len(first_name))
print(len(last_name))
print(len(full_name))
```

```python
# String formatting 
# % operator is used to fomat a set of variables enclosed in a tuple (a fixed size list), together with a formating string, which contains normal text together with "argument specifiers", special symbols lik "%s", "%d", "%f", "%.number of digitsf"
```

```python
"""
    %s - String (or any object with a string representation, like numbers)
    %d - Integers
    %f - Floating point numbers
    "%.number of digitsf" - Floating point numbers with fixed precision    
    
"""
```

```python
# Strings only
first_name = 'Peter'
last_name = 'Garcia'
language = 'Python'
formated_string = 'I am %s %s. I teach %s' %(first_name, last_name, language)
print(formated_string)
```

```python
# Strings  and numbers
radius = 10
pi = 3.14
area = pi * radius ** 2
formated_string = 'The area of circle with a radius %d is %.2f.' %(radius, area) # 2 refers the 2 significant digits after the point
print(formated_string)
```

```python
python_libraries = ['Django', 'Flask', 'NumPy', 'Matplotlib','Pandas']
formated_string = 'The following are python libraries:%s' % (python_libraries)
print(formated_string) 
```

```python
# New style string formatting (str.format)
first_name = 'Peter'
last_name = 'Garcia'
language = 'Python'
formated_string = 'I am {} {}. I teach {}'.format(first_name, last_name, language)
print(formated_string)
```

```python
a = 4
b = 3

print('{} + {} = {}'.format(a, b, a + b))
print('{} - {} = {}'.format(a, b, a - b))
print('{} * {} = {}'.format(a, b, a * b))
print('{} / {} = {:.2f}'.format(a, b, a / b)) # limits it to two digits after decimal
print('{} % {} = {}'.format(a, b, a % b))
print('{} // {} = {}'.format(a, b, a // b))
print('{} ** {} = {}'.format(a, b, a ** b))
```

```python
# String interpolation /f-String
a = 4
b = 3
print(f'{a} + {b} = {a +b}')
print(type(a))
```

```python
# Python strings as sequence of characters. 
```

```python
# Simplest way of extracting single characters from strings 
language = 'Python'
a,b,c,d,e,f = language # unpacking sequence characters into variables
print(a) # P
print(b) # y
print(c) # t
print(d) # h
print(e) # o
print(f) # n
```

```python
# Accessing Characters in Strings by Index 
# couting starts from zero. First letter of a string is at zero index  and last letter of the string is the lenght of a string minus one. 
language = 'Python'
first_letter = language[0]
print(first_letter)
last_index = len(language) - 1
print(last_index)
last_letter = language[last_index]
print(last_letter)
```

```python
# If we want to start from the right end we can use negative indexing. -1 is the last index. 

language = 'Python'
last_letter = language[-1]
print(last_letter)

# Second last letter 
second_last = language[-2]
print(second_last)
```

```python
# Slicing Python Strings 
# slice strings into substrings 
language = 'Python'
first_three = language[0:3] # Starts at zero index and up to 3 but not including three
print(first_three)

# Last 3 letters
last_three = language[3:6]
print(last_three)

last_three = language[3:]
print(last_three)
```

```python
# Reversing a string 
# We can reverse strings in python by: 
greetings = "Hello World"
print(greetings[::-1])
```

```python
# Skipping characters while slicing 
# it is possible to skip characters while slicing by passing step argument to slice method. 
language = 'Python'
pto = language[0:6:2] 
print(pto)
```

```python
# Capitalize 
challenge = 'thirty days of python'
print(challenge.capitalize())
```

```python
# Count 
print(challenge.count('y'))
print(challenge.count('y', 7, 16))
print(challenge[16])
```

```python
# endswith(): checks if a string ends with a specific ending 
print(challenge.endswith('on'))
print(challenge.endswith('  python'))
print(challenge.endswith(' python'))
print(challenge.endswith('python'))
```

```python
#expandstabs(): replaces tab characters with spaces, default tab size is 8. It takes tab size argument. 
challenge = 'thirty\tdays\tof\tpython'
print(challenge)
print(challenge.expandtabs())
print(challenge.expandtabs(1))
print(challenge.expandtabs(10))
```

```python
# find(): returns the index of the first occurrence of a substring. if not found returns -1
challenge = 'thirty days of python'
print(challenge.find('y'))
print(challenge.find('on'))
print(challenge.find('th'))
```

```python
#rfind(): Return the index of the last occurrence of a substring. if not found returns -1. 
challenge = 'thirty days of python'
print(challenge.rfind('y'))
print(challenge.rfind('on'))
```

```python
# index(): returns the  lowest index of a substring, additional arguments indicate starting and ending index. 
challenge = 'thirty days of python'
sub_string = 'da'
print(challenge.index(sub_string))  # 7
print(challenge.index(sub_string, 8)) # error
```

```python
# rindex(): Returns the highest index of a substring, additional arguments indicate starting and ending index (default 0 and string length - 1)
challenge = 'thirty days of python'
sub_string = 'da'
print(challenge.rindex(sub_string))  # 7
print(challenge.rindex(sub_string, 9)) # error
print(challenge.rindex('on', 8)) # 19
```

```python
# isalnum(): Checks alphanumeric character
challenge = 'ThirtyDaysPython'
print(challenge.isalnum()) # True

challenge = '30DaysPython'
print(challenge.isalnum()) # True

challenge = 'thirty days of python'
print(challenge.isalnum()) # False, space is not an alphanumeric character

challenge = 'thirty days of python 2019'
print(challenge.isalnum()) # False
```

```python
# join() return a concatenated string 
web_tech = ['HTML', 'CSS', 'JavaScript', 'React']
result = ' '.join(web_tech)
print(result) # 'HTML CSS JavaScript React'
```

```python
web_tech = ['HTML', 'CSS', 'JavaScript', 'React']
result = '# '.join(web_tech)
print(result) # 'HTML# CSS# JavaScript# React'
```

```python
# strip(): Removes all given characters starting from the beginning and end of the string
challenge = 'thirty days of pythoonnn'
print(challenge.strip('noth')) # 'irty days of py'
```

```python
# replace(): Replaces substring with a given string
challenge = 'thirty days of python'
print(challenge.replace('python', 'coding')) # 'thirty days of coding'
```

```python
# split(): Splits the string, using given string or space as a separator
challenge = 'thirty days of python'
print(challenge.split()) # ['thirty', 'days', 'of', 'python']
challenge = 'thirty, days, of, python'
print(challenge.split(', ')) # ['thirty', 'days', 'of', 'python']
```

```python
# title(): Returns a title cased string
challenge = 'thirty days of python'
print(challenge.title()) # Thirty Days Of Python
```

```python
# swapcase(): Converts all uppercase characters to lowercase and all lowercase characters to uppercase characters
challenge = 'thirty days of python'
print(challenge.swapcase())   # THIRTY DAYS OF PYTHON
challenge = 'Thirty Days Of Python'
print(challenge.swapcase())  # tHIRTY dAYS oF pYTHON
```

```python
# startswith(): Checks if String Starts with the Specified String

challenge = 'thirty days of python'
print(challenge.startswith('thirty')) # True

challenge = '30 days of python'
print(challenge.startswith('thirty')) # False
```
